## Part 6: Construct the Minimum Risk Portfolio

Compute the minimum risk portfolio based on your previously estimated risks.

```python
# YOUR CHANGES HERE

covariance_matrix = pd.read_csv(
    "estimated_covariance.tsv",
    sep="\t",
    index_col=0
)

tickers = ["AAPL", "SPY", "TSLA", "LMT", "JPM", "XOM"]
```

```python
def portfolio_variance(weights, covariance_matrix):
    return weights @ covariance_matrix.values @ weights
```

```python
import numpy as np

n_assets = len(tickers)

initial_weights = np.repeat(1 / n_assets, n_assets)

constraints = {
    "type": "eq",
    "fun": lambda weights: np.sum(weights) - 1
}

bounds = [(0, 1) for _ in range(n_assets)]
```

```python
from scipy.optimize import minimize

result = minimize(
    portfolio_variance,
    initial_weights,
    args=(covariance_matrix.loc[tickers, tickers],),
    method="SLSQP",
    bounds=bounds,
    constraints=constraints
)
```

```python
minimum_risk = pd.DataFrame({
    "asset": tickers,
    "allocation": result.x
})

print("Optimization successful:", result.success)
print("Allocation sum:", minimum_risk["allocation"].sum())

minimum_risk
```

## Part 7: Build Efficient Frontier Portfolios

Compute 101 portfolios along the mean-variance efficient frontier with evenly spaced estimated returns.
The first portfolio should be the minimum risk portfolio from part 4, and the last portfolio should be the maximum return portfolio from part 3.
The estimated return of each portfolio should be higher than the previous by one percent of the difference between the first and last portfolios.
That is, the estimated return of the portfolios should be similar to `np.linspace(min_risk_return, max_return, 101)`.


```python
# YOUR CHANGES HERE
import pandas as pd
import numpy as np
from scipy.optimize import minimize

estimated_returns = pd.read_csv(
    "estimated_returns.tsv",
    sep="\t"
).set_index("asset")["estimated_return"]

covariance_matrix = pd.read_csv(
    "estimated_covariance.tsv",
    sep="\t",
    index_col=0
)

minimum_risk = pd.read_csv(
    "minimum_risk.tsv",
    sep="\t"
)

maximum_return = pd.read_csv(
    "maximum_return.tsv",
    sep="\t"
)

tickers = ["AAPL", "SPY", "TSLA", "LMT", "JPM", "XOM"]
```

```python
min_weights = minimum_risk.set_index("asset").loc[tickers, "allocation"].values
max_weights = maximum_return.set_index("asset").loc[tickers, "allocation"].values

mean_returns = estimated_returns.loc[tickers].values

min_risk_return = min_weights @ mean_returns
max_return = max_weights @ mean_returns

print("Minimum-risk portfolio return:", min_risk_return)
print("Maximum-return portfolio return:", max_return)
```

```python
target_returns = np.linspace(
    min_risk_return,
    max_return,
    101
)

target_returns[:5], target_returns[-5:]
```

```python
cov_matrix = covariance_matrix.loc[tickers, tickers].values

def portfolio_variance(weights):
    return weights @ cov_matrix @ weights
```

```python
frontier_portfolios = []

initial_weights = min_weights.copy()

bounds = [(0, 1) for _ in tickers]

for target_return in target_returns:

    constraints = [
        {
            "type": "eq",
            "fun": lambda weights: np.sum(weights) - 1
        },
        {
            "type": "eq",
            "fun": lambda weights, target=target_return:
                weights @ mean_returns - target
        }
    ]

    result = minimize(
        portfolio_variance,
        initial_weights,
        method="SLSQP",
        bounds=bounds,
        constraints=constraints
    )

    if not result.success:
        print("Optimization failed:", target_return, result.message)

    frontier_portfolios.append(result.x)

    # Use current solution as next starting point
    initial_weights = result.x
```

```python
efficient_frontier = pd.DataFrame(
    frontier_portfolios,
    columns=tickers
)

efficient_frontier.insert(
    0,
    "estimated_return",
    target_returns
)

efficient_frontier
```

```python
print("Number of portfolios:", len(efficient_frontier))

print(
    "Allocation sums:",
    efficient_frontier[tickers].sum(axis=1).min(),
    "to",
    efficient_frontier[tickers].sum(axis=1).max()
)

print(
    "First return:",
    efficient_frontier["estimated_return"].iloc[0]
)

print(
    "Last return:",
    efficient_frontier["estimated_return"].iloc[-1]
)
```

```python
return_steps = np.diff(
    efficient_frontier["estimated_return"]
)

print("Return step:", return_steps[0])
print("All steps equal:", np.allclose(return_steps, return_steps[0]))
```

```python
# YOUR CHANGES HERE
frontier_risks = []

for weights in frontier_portfolios:
    variance = weights @ cov_matrix @ weights
    standard_deviation = np.sqrt(variance)
    frontier_risks.append(standard_deviation)
```

## Part 8: Check Maximum Return Portfolio Stability

Check the stability of the maximum return portfolio by resampling the estimated risk/return model.

Repeat 1000 times -
1. Use `np.random.multivariate_normal` to generate 23 return samples using your previously estimated risks and returns.
2. Estimate the return of each asset using that resampled return history.
3. Check which asset had the highest return in those resampled estimates.

This procedure is a reduced and simplified version of the Michaud resampled efficient frontier procedure that takes uncertainty in the risk model into account.

```python
# YOUR CHANGES HERE
tickers = ["AAPL", "SPY", "TSLA", "LMT", "JPM", "XOM"]

estimated_returns = pd.read_csv(
    "estimated_returns.tsv",
    sep="\t"
).set_index("asset")["estimated_return"]

covariance_matrix = pd.read_csv(
    "estimated_covariance.tsv",
    sep="\t",
    index_col=0
)

mean_vector = estimated_returns.loc[tickers].values
cov_matrix = covariance_matrix.loc[tickers, tickers].values
```

```python
np.random.seed(42)

winner_counts = {ticker: 0 for ticker in tickers}

for _ in range(1000):

    simulated_returns = np.random.multivariate_normal(
        mean=mean_vector,
        cov=cov_matrix,
        size=23
    )

    resampled_means = simulated_returns.mean(axis=0)

    winner_index = np.argmax(resampled_means)
    winner_ticker = tickers[winner_index]

    winner_counts[winner_ticker] += 1
```

```python
stability_results = pd.DataFrame({
    "asset": tickers,
    "times_highest_return": [winner_counts[t] for t in tickers]
})

stability_results["proportion"] = (
    stability_results["times_highest_return"] / 1000
)

stability_results
```

```python
print("Total simulations:", stability_results["times_highest_return"].sum())
```

Save a file "max_return_probabilities.tsv" with the distribution of highest return assets.
The header row should have columns "asset" and "probability".
There should be a data row for each asset and its sample probability of having the highest return based on those 1000 resampled estimates.


```python
# YOUR CHANGES HERE

max_return_probabilities = pd.DataFrame({
    "asset": tickers,
    "probability": [winner_counts[ticker] / 1000 for ticker in tickers]
})

max_return_probabilities
```
